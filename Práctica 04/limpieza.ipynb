```python
import pandas as pd
```

```python
df = pd.read_csv('data/infracciones.csv')
print(df.head())
```

```python
df.info()
```

```python
df.describe()
```

```python
columnas_constantes = [col for col in df.columns if df[col].nunique() <= 1]
columnas_constantes
```

```python
# Encontrar filas duplicadas
filas_duplicadas = df[df.duplicated()]
print(f"Filas duplicadas encontradas: {len(filas_duplicadas)}")

# Mostrar las filas duplicadas
print(filas_duplicadas)
```

```python
df[df['MONTO_INFRACCION'] > 0]
```

```python
before = len(df)
df = df.drop_duplicates()
after = before - len(df)
print(f"Se eliminaron {after} filas duplicadas.")
```

```python
# Filas con valores vacíos
filas_con_nan = df[df.isnull().any(axis=1)]
print(f"Filas con valores vacíos: {len(filas_con_nan)}")
print(filas_con_nan)
```

```python
# Valores de la columna SE_PAGO_MULTAS
valores_se_pago_multas = df['SE_PAGO_MULTAS'].unique()
print(f"Valores únicos en SE_PAGO_MULTAS: {valores_se_pago_multas}")
```

```python
# Exportamos el df limpio
df.to_csv('data/infracciones_limpio.csv', index=False)
```
